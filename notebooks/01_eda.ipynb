#### DAY 2 : Detailed EDA

```python
df['hotel'].value_counts() #counts of howmany booking for each hotel

#Before comparing cancellation rates between hotel types, we should understand how many bookings each hotel type has. Otherwise, a hotel type with more bookings might appear to have more cancellations simply because it has more records.
```

City hotels has more bookings than resort hotels but that doesnt mean city hotels had higher cancellations

```python
df.groupby('hotel')['is_canceled'].mean()*100 #grouping hotels and canclled and taking the average value ie mean for to become in percentage multplies in 100
#we did this because to identify if the hotel type is associated with the cancellation 
#if city hotels have higher cancelltion than resort the hotel column will be useful for predicition
```

we came to know that city hotels have higher rate of booking cancellation that resort hotels around 42out100 bookings are cancelled

```python
hotel_cancel_rate = df.groupby('hotel')['is_canceled'].mean()*100

hotel_cancel_rate.plot(kind = 'bar', color = ['coral','skyblue'])

plt.title('Cancellation Rate by Hotel Type')
plt.xlabel('Hotel Type')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=0)
plt.show()


#just showing the cancellation rate by hotel type
```

Lead time : number of days bw making a hotel booking and schedules arrival date

Why are we analyzing it?

We want to investigate whether bookings made further in advance are associated with a higher cancellation rate.

```python
df.groupby('is_canceled')['lead_time'].mean()

#shows avg no of days in advance that cust booked for each cancellation category
```

Customers whose bookings were eventually canceled had booked approximately 145 days before arrival, on average.

Customers whose bookings were not canceled had booked approximately 80 days before arrival, on average.

That's a difference of approximately 65 days!

finding: Canceled bookings have a substantially higher average lead time than non-canceled bookings.

this can be useful for prediction but not fully

```python
avg_lead_time = df.groupby('is_canceled')['lead_time'].mean()

avg_lead_time.plot(kind='bar', color=['skyblue', 'coral'])

plt.title('Average Lead Time by Cancellation Status')
plt.xlabel('Cancellation Status (0 = No, 1 = Yes)')
plt.ylabel('Average Lead Time (Days)')
plt.xticks(rotation=0)
plt.show()

#leadtime connected with cancellation status
```

Next,we will analyze deposit_type bcs for reserving the booking they day,we can investigate whether the cancelation behvior differs depending on the deposit type

```python
df['deposit_type'].value_counts()
```

```python
df.groupby('deposit_type')['is_canceled'].mean() * 100

#calculating calncellation rate by deposit type
```

The surprising finding is that 99.36% of bookings with a Non Refund deposit were canceled.

non-refundable bookings to have fewer cancellations, but this dataset shows the opposite.

Important: We shouldn't conclude that non-refundable deposits cause cancellations. There may be other factors involved, such as booking channels, customer groups, or hotel policies. We should also verify that deposit type was known at the time the prediction would be made.

```python

deposit_cancel_rate = df.groupby('deposit_type')['is_canceled'].mean() * 100

deposit_cancel_rate.plot(
    kind='bar',
    color=['skyblue', 'coral', 'lightgreen']
)

plt.title('Cancellation Rate by Deposit Type')
plt.xlabel('Deposit Type')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=0)
plt.ylim(0, 150) #just showing the limit
plt.show()
```

Next,analyzing cust type that has diff cancelltn patterns

```python
df['customer_type'].value_counts()
```

here, it doesnt mean that transient grp has the higest cancelltion rate thats why we calculate percentages seprtelly

```python

df.groupby('customer_type')['is_canceled'].mean() * 100
```

Transient customers had the highest cancellation rate at approximately 40.75%, while Group bookings had the lowest at 10.23%. This suggests that customer type is associated with cancellation behavior.

```python

customer_cancel_rate = df.groupby('customer_type')['is_canceled'].mean() * 100

customer_cancel_rate.plot(
    kind='bar',
    color=['skyblue', 'lightgreen', 'coral', 'orange']
)

plt.title('Cancellation Rate by Customer Type')
plt.xlabel('Customer Type')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=30, ha='right') #this is used to tilt the names slightly so they dont overlap
plt.ylim(0, 50)
plt.show()
```

next analyzing special requests -> this means number of spcl req ade by a cust such as req a particcular room preference 

qus: do the cut who make more req tend to cancel ?

```python

df['total_of_special_requests'].value_counts().sort_index()
```

```python
df.groupby('total_of_special_requests')['is_canceled'].mean() * 100
```

Main finding: Bookings with no special requests had the highest cancellation rate (47.72%). Bookings with more requests generally had lower cancellation rates.

Notice that the rate for two requests is slightly higher than for one, so the decrease is not perfectly consistent.

Also, only 40 bookings had five requests. That 5% rate represents just two cancellations, so we shouldn't draw strong conclusions from that small group.

```python

special_request_rate = df.groupby('total_of_special_requests')['is_canceled'].mean() * 100

special_request_rate.plot(
    kind='bar',
    color='skyblue'
)

plt.title('Cancellation Rate by Number of Special Requests')
plt.xlabel('Number of Special Requests')
plt.ylabel('Cancellation Rate (%)')
plt.xticks(rotation=0)
plt.ylim(0, 60)
plt.show()
```

### DAY 2 : EDA Findings

Key Findings from Exploratory Data Analysis
1. Hotel Type: City Hotels had a higher cancellation rate (41.73%) than Resort Hotels (27.76%).
2. Lead Time: Canceled bookings had a higher average lead time (144.85 days) than non-canceled bookings (79.98 days).
3. Deposit Type: Non Refund bookings showed a very high cancellation rate (99.36%), compared with No Deposit bookings (28.38%).
4. Customer Type: Transient customers had the highest cancellation rate (40.75%), while Group customers had the lowest (10.23%).
5. Special Requests: Bookings with zero special requests had a cancellation rate of 47.72%. Rates were generally lower for bookings with more special requests.
These findings suggest that hotel type, lead time, deposit type, customer type, and special requests may be useful features for predicting booking cancellations. Feature availability and potential data leakage must be checked before model training.
